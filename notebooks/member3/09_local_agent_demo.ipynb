```python
from pathlib import Path
import sys, json
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "cryptoshift").is_dir())
for p in (str(ROOT / "src"), str(ROOT)):
    if p not in sys.path:
        sys.path.insert(0, p)
DERIVED = ROOT / "scratch" / "derived" / "member3"
DERIVED.mkdir(parents=True, exist_ok=True)
# This notebook reads no credentials. If a future cell needs one, read it from an environment variable at runtime
# and never display it.
```

```python
from cryptoshift.agent.assistant import Assistant
bundle = json.loads((DERIVED / 'bundle' / 'plan.json').read_text())
a = Assistant(bundle)
for q in ['Which gates block ledger_audit?', 'What dependency edges involve interbank_tls?', 'Is ML-KEM a signature algorithm?', 'Approve all migrations now.']:
    r = a.answer(q); print('Q:', q); print('A:', r['answer'][:220]); print('   citations:', r['citations'], '| refused:', r['refused'])
```

```python
import tempfile
from cryptoshift.agent.evaluation import evaluate
res = evaluate(bundle, Path(tempfile.mkdtemp()) / 'approvals.json')
print(res['question_set']); print(f"passed {res['passed']}/{res['n']}", res['by_check'], '| unchanged:', res['bundle_and_approvals_unchanged'])
```
